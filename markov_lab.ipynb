```python
import numpy as np
rng = np.random.default_rng(0)
```

Initialise random 3x3 matrix, normalising rows to sum to 1.

```python
Q = rng.random((3,3))
Q = Q / Q.sum(axis=1)[:, None]
```

# 1: stationary distribution via right-eigenvectors
sQ = s: the stationary distribution is a left-eigenvector of Q.
linalg.eig finds right-eigenvectors. Finding the right-eigenvectors of Q.T is equivalent to finding left-eigenvectors of Q.

```python
vals, vecs = np.linalg.eig(Q.T) 
k = np.argmin(np.abs(vals - 1)) 
s = np.real(vecs[:, k])
s = s / s.sum()
s
```

```python
np.allclose(s @ Q, s) # just to check if the right-eigenvector is truly the stationary distribution
```

# 2: repeated steps
In a random simulation through a Markov chain, as the number of steps tends to infinity, the proportions of time spent in each state will tend to the stationary distribution. Here, we iterate from 1 to 100,000 steps. Note how the summed absolute difference between the empirical time spent in each state and the real stationary distribution tends to 0 as the number of steps increases.

```python
step_counts = [1, 10, 100, 1_000, 10_000, 100_000]
for n_steps in step_counts:
    states = np.empty(n_steps, dtype=int)
    states[0] = 0
    for t in range(1, n_steps):
        states[t] = rng.choice(3, p=Q[states[t-1]])
    empirical = np.bincount(states, minlength=3) / n_steps
    total_diff = np.sum(np.abs(s - empirical))
    print(f"num steps: {n_steps}, time spent in each state: {empirical}, summed difference with s: {total_diff}")

print(f"actual s: {s}")
```

# 3: repeated powers of Q 
Repeated multiplication of Q with itself causes the rows to tend to the stationary distribution (as n tends to infinity).

```python
for n in [1, 5, 10, 50, 100, 1000, 10000]:
    q_to_n = np.linalg.matrix_power(Q, n)
    print(f" Q^{n}: {q_to_n}")

print(f"actual s: {s}")
```
