```python
import numpy as np
import matplotlib.pyplot as plt
rng = np.random.default_rng()
```

Taking 10000 samples from 5 multivariate normals
Each iterates from 2 to -2 on the off diagonal of the covariance matrix


```python
xy_covs = [2, 1, 0, -1, -2]

for c in xy_covs:
    cov = np.array([[1, c], [c, 5]])
    samples = np.random.multivariate_normal([2, 5], cov, 10000)
    
    plt.scatter(samples[:, 0], samples[:, 1])
    
    eigenv, eigenvec = np.linalg.eig(cov)
    
    for vec in eigenvec:
        plt.quiver(*[0,0], *vec, color=['r'], scale=5)
    plt.show()

    print(np.cov(samples.T))
```
